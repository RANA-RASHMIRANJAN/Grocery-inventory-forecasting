```python
import pandas as pd

df = pd.read_csv('/content/retail_demand_2025.csv')

print(df.shape)
print(df.columns.tolist())
```

```python
# Target variable
y = df['Units_Sold']

# Features
X = df.drop(['Units_Sold', 'Sales', 'Date'], axis=1)

print("X shape:", X.shape)
print("y shape:", y.shape)
print("Features:")
print(X.columns.tolist())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

categorical_cols = [
    'Store_ID',
    'City',
    'Product_ID',
    'Product',
    'Category'
]

numeric_cols = [
    'Price',
    'Discount',
    'Promotion',
    'Holiday',
    'Stock'
]

preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols),
        ('num', 'passthrough', numeric_cols)
    ]
)

X_encoded = preprocessor.fit_transform(X)

print("Encoded data shape:", X_encoded.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split

# Categorical columns
categorical_cols = ['Store_ID', 'City', 'Product_ID', 'Product', 'Category']

# Numerical columns
numeric_cols = ['Price', 'Discount', 'Promotion', 'Holiday', 'Stock']

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols),
        ('num', 'passthrough', numeric_cols)
    ]
)

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

# Encode
X_train_encoded = preprocessor.fit_transform(X_train)
X_test_encoded = preprocessor.transform(X_test)

print("Training data:", X_train_encoded.shape)
print("Testing data:", X_test_encoded.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train_encoded, y_train)

print("Model training completed!")
```

```python
y_pred = model.predict(X_test_encoded)

print("First 10 Actual Values:")
print(y_test.iloc[:10].values)

print("\nFirst 10 Predicted Values:")
print(y_pred[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("MAE :", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("R² Score:", round(r2, 4))
print("R² Score (%):", round(r2 * 100, 2), "%")
```
